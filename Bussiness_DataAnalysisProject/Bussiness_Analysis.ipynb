## Business Action Plan: What the Business Should Do
### Based on complete analysis of Task 1 to Task 20

This section turns all the negative findings from the analysis into clear, practical recommendations. Every point below is based on real numbers found in the data.

---

### 1. Fix the Furniture Category
Furniture has almost the same sales as Office Supplies (**741,999** vs **719,047**) but its profit is only **18,451**, compared to Office Supplies' **122,490**. This means Furniture is selling well but barely making money.

**What to do:** Review pricing and supplier costs for Furniture. Reduce or stop discounts on Furniture products specifically, since this category cannot absorb heavy discounting like Technology can.

---

### 2. Stop Losses in Bookcases and Tables
Bookcases sub-category has a **negative profit of 3,472.56** even with sales of **114,879**. Tables and Supplies sub-categories also show a **negative profit margin** in the margin analysis.

**What to do:** Either renegotiate supplier costs for Bookcases and Tables, raise their prices slightly, or reduce discount eligibility on these specific sub-categories. If margins cannot be fixed, consider phasing out the worst performing models within these sub-categories.

---

### 3. Control Discounting, Especially Above 20 Percent
The scatter plot and correlation both confirm discount is hurting profit, with a correlation of **negative 0.22**. Profit regularly turns negative once discount crosses **around 0.3 to 0.4**, and losses get worse at **0.6 to 0.8** discount levels.

**What to do:** Set a maximum discount cap, for example **20 to 25 percent**, unless it is a special clearance sale approved separately. Sales team should not have authority to give discounts above this cap without manager approval.

---

### 4. Discontinue or Renegotiate the Worst Loss Making Products
These 15 products are draining profit the most:
- Cubify CubeX 3D Printer Double Head Print: losing **8,879.97**
- Lexmark MX611dhe Monochrome Laser Printer: losing **4,589.97**
- Cubify CubeX 3D Printer Triple Head Print: losing **3,839.99**
- Chromcraft Bull-Nose Wood Oval Conference Tables and Bases: losing **2,876.12**
- Bush Advantage Collection Racetrack Conference Table: losing **1,934.40**
- GBC DocuBind P400 Electric Binding System: losing **1,878.17**
- Cisco TelePresence System EX90 Videoconferencing Unit: losing **1,811.08**
- Martin Yale Chadless Opener Electric Letter Opener: losing **1,299.18**
- Balt Solid Wood Round Tables: losing **1,201.06**
- BoxOffice By Design Meeting Room Tables: losing **1,148.44**
- Riverside Furniture Oval Coffee Table set: losing **1,147.40**
- Epson TM-T88V Direct Thermal Printer: losing **1,057.23**
- Hon 2090 Pillow Soft Series Chairs: losing **989.05**
- O'Sullivan 4-Shelf Bookcase: losing **975.10**
- Bretford Height-Adjustable Work Tables: losing **964.19**

**What to do:** Both Cubify CubeX 3D Printers are losing the most money by far and should be reviewed first, either by raising the price, cutting the discount offered on them, or dropping them from the catalog if losses continue. The pattern shows large printers and large conference tables are consistently unprofitable, likely due to high shipping or discount costs on big-ticket items. Apply the same review to the rest of the list.

---

### 5. Turn Around the Weakest Cities
Twelve cities are currently losing money, led by:
- Philadelphia: losing **13,837.77**
- Houston: losing about **10,000**
- San Antonio: losing about **8,000**
- Lancaster, Chicago, Burlington, Dallas, Phoenix, Aurora, Jacksonville, Memphis, and Louisville: all showing smaller losses

**What to do:** Philadelphia needs the most urgent attention since it has good sales but the worst profit loss of any city, which usually means heavy discounting is happening there specifically. Check discount levels given to customers in Philadelphia and Houston first. For the other cities with smaller losses, monitor for now but do not expand marketing spend there until profit improves.

---

### 6. Protect and Grow the Profitable Cities
New York City is the strongest city by far with profit of **62,036**, followed by Los Angeles at **30,440** and Seattle at **29,156**.

**What to do:** Increase marketing and inventory investment in these three cities since they already prove they can convert sales into strong profit. Consider studying what is different about customer behavior or product mix here compared to Philadelphia, and try to replicate it in weaker cities.

---

### 7. Keep Standard Class Shipping as the Core, but Review Same Day
Standard Class brings in the most profit at **164,088** and is used in **59.7 percent** of orders, showing it is the most efficient shipping mode. Same Day is the least used and brings the least profit at only **15,891**.

**What to do:** Keep promoting Standard Class as the default option since it is already working well. Same Day shipping should be reviewed to check if its higher cost is worth offering, or if it should only be available for premium priced orders.

---

### 8. Focus Marketing on Consumer and Office Supplies
The most common transaction type by far is Consumer segment buying Office Supplies, with **3,127 transactions**, much higher than any other combination.

**What to do:** Since this is the core of the business, keep Office Supplies stock levels high and consider loyalty offers targeted at Consumer segment customers specifically, since they are already the most engaged group.

---

### 9. Promote High Margin Sub-Categories More
Labels, Paper, Envelopes, Copiers, and Fasteners have the healthiest profit margins, all above **30 percent**.

**What to do:** Feature these products more in promotions and bundle deals, since they make the most profit per dollar of sales. Pushing more volume into these sub-categories, instead of low margin ones like Tables or Bookcases, will improve overall company profit without needing to increase total sales.

---

### 10. Reward and Retain Top Customers
A small group of top customers, led by **SM-20320**, **TC-20980**, **RB-19360**, and **TA-21385**, are generating the highest individual sales.

**What to do:** Set up a loyalty or account manager program for these top 15 customers so they do not switch to a competitor. Losing even one of them could have a bigger impact than losing many smaller customers combined.

---

### Summary of Priority Actions
1. Cap discounts, especially in Philadelphia and Houston
2. Review or discontinue the two Cubify CubeX 3D Printer models first
3. Fix pricing or costs in Bookcases and Tables sub-categories
4. Invest more in New York City, Los Angeles, and Seattle
5. Push more sales toward Labels, Paper, Envelopes, and Copiers
6. Set up retention offers for the top 15 customers

These six actions directly target the biggest sources of lost profit found across all twenty analysis tasks, and should have the fastest visible impact on the business's overall profitability.
