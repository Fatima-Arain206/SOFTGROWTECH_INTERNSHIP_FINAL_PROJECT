```python
# import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
# data load
df = pd.read_csv(r'/home/fatima/Desktop/SOFTGROWTECH_INTERNSHIP_FINAL_PROJECT/Bussiness_DataAnalysisProject/Super_.csv',encoding='latin1')
```

```python
df.head()
```

```python
# over all profit and sales analysis
print(df['Sales'].sum())
print(df['Profit'].sum())
```

```python
# check profit and sales by catagory data
df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
```

```python
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('Category')[['Sales','Profit']].sum()
```

```python
df.groupby('Category')[['Sales','Profit']].sum().plot(kind='bar')
plt.title('Sales and Profit by Category')
plt.xticks(rotation=0)
plt.xlabel('Product Category')
plt.ylabel('Sales and profit')
plt.show()
```

```python
# check with sub category
df.groupby('Sub-Category')[['Sales','Profit']].sum()
```

```python
# sub cat sale
colors = ['#FF5733', '#33FF57', '#3357FF', '#F1C40F', '#8E44AD', '#E74C3C', '#2ECC71', '#3498DB', '#F39C12', '#1ABC9C']
df.groupby('Sub-Category')['Sales'].sum().plot(kind='barh', color=colors)
plt.title('Sales by Sub-Category')
plt.xlabel('Sales')
plt.ylabel('Sub-Category Items')
plt.show()
```

```python
# sub category by pro
df.groupby('Sub-Category')['Profit'].sum().plot(kind='barh',
                        color=colors                        )
plt.title('Profit by Sub Category')
plt.xlabel('profit')
plt.ylabel('Sub Category Items')
plt.show()
```

```python
# Region wise sales
df.groupby('Region')[['Sales','Profit']].sum()
```

```python
# graph
df.groupby('Region')['Sales'].sum().plot(kind='pie',labels=['Central','East','South','West'],autopct='%1.1f%%')
plt.title('Regional Sales Percentage')
plt.show()
```

```python
df.groupby('Region')['Profit'].sum().plot(kind='pie',labels=['Central','East','South','West'],autopct='%1.1f%%')
plt.title('Regional Profit Percentage')
plt.show()
```
