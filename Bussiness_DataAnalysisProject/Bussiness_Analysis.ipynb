## Business Data Analysis Summary 

This section covers the first ten analysis steps performed on the Superstore dataset. Each step looks at a different angle of the business performance.

## **1. Overall Sales and Profit**
Total Sales came out to **2,297,200.86** and Total Profit came out to **286,397.02**. This gives a baseline of how the whole business is performing.

## **2. Category wise Performance**
Technology had the highest sales at **836,154.03** and also the highest profit at **145,454.95**. Office Supplies had lower sales of **719,047.03** but still gave a healthy profit of **122,490.80**. Furniture had almost the same sales as Office Supplies at **741,999.80**, but the profit was very low at only **18,451.27**, showing this category is not very efficient.

## **3. Sub Category wise Performance**
Chairs had the highest sales at **328,449.10**. Copiers gave the best profit at **55,617.82** even with moderate sales. Bookcases were a loss making sub category with profit of **negative 3,472.56**, which needs attention.

## **4. Region wise Performance**
West region had the highest profit at **108,418.45** with sales of **725,457.82**. East region had the highest sales at **678,781.24** with profit of **91,522.78**. Central region was the weakest with sales of only **501,239.89** and the lowest profit of **39,706.36**.

 ## **5. Segment wise Performance**
Consumer segment dominates with **50.6 percent** of total sales. Corporate segment comes next with **30.7 percent**, and Home Office is the smallest at **18.7 percent** of sales.

## **6. Monthly Sales Trend**
Sales peak in **November** at **352,461.07**, followed by **December** at **325,293.50**. The lowest sales month is **February** at only **59,751.25**, which shows a strong seasonal pattern around the year end holidays.

## **7. Top Products by Sales and Profit**
The **Canon imageCLASS 2200 Advanced Copier** is the top performer, with sales of **61,599.82** and profit of **25,199.93**. Products like binding machines, copiers, and printers dominate the top profitable products list.

## **8. Yearly Sales Trend**
Sales were highest in **2017** at close to **720,000**, followed by **2016** at around **590,000**. 2014 and 2015 were lower, close to **450,000** and **420,000**. This shows the business grew steadily over the years.

## **9. Daily Sales Trend**
The daily sales trend line shows a lot of ups and downs across the years, with several sharp spikes reaching **25,000** to **28,000** on specific days. This is normal for order level data and shows sales are not steady day to day.

## **10. City wise Performance**
**New York City** is the clear leader with sales of **256,368.16** and profit of **62,036.98**. **Los Angeles** follows with sales of **175,851.34** and profit of **30,440.76**. Interestingly, **Philadelphia** had good sales of **109,077.01** but a negative profit of **negative 13,837.77**, meaning heavy discounts or losses are happening there.

### Key Takeaways so far
Technology is the strongest category, Furniture needs improvement, West region and Consumer segment drive the most business, November is the peak sales month, and some cities like Philadelphia show sales without profit, which is worth investigating further in the next set of tasks.
